```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/Q05_property_listings.csv')

print("Dataset loaded successfully")
print("Rows and Columns:", df.shape)

display(df.head())
```

```python
print("Column Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate listings:", df.duplicated(subset='listing_id').sum())

df = df.drop_duplicates(subset='listing_id')

print("Duplicates after cleaning:",
      df.duplicated(subset='listing_id').sum())
```

```python
df['neighbourhood'] = df['neighbourhood'].fillna(
    df['neighbourhood'].mode()[0]
)

print("Missing neighbourhood values:",
      df['neighbourhood'].isnull().sum())
```

```python
df['listing_price_inr'] = df['listing_price_inr'].fillna(
    df['listing_price_inr'].median()
)

print("Missing price values:",
      df['listing_price_inr'].isnull().sum())
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nFinal dataset shape:", df.shape)

display(df.head())
```

```python
df['size_group'] = pd.cut(
    df['property_area_sqft'],
    bins=[0, 1000, 1500, 2000, 2500, np.inf],
    labels=[
        'Small',
        'Medium',
        'Large',
        'Very Large',
        'Luxury'
    ]
)

display(
    df[['property_area_sqft', 'size_group']].head()
)
```

```python
df['age_group'] = pd.cut(
    df['property_age_years'],
    bins=[-1, 5, 10, 20, 30, np.inf],
    labels=[
        '0-5 Years',
        '6-10 Years',
        '11-20 Years',
        '21-30 Years',
        '31+ Years'
    ]
)

display(
    df[['property_age_years', 'age_group']].head()
)
```

```python
df['price_per_sqft'] = (
    df['listing_price_inr'] /
    df['property_area_sqft']
)

display(
    df[
        [
            'listing_id',
            'property_area_sqft',
            'listing_price_inr',
            'price_per_sqft'
        ]
    ].head()
)
```

```python
neighbourhood_analysis = (
    df.groupby('neighbourhood')['listing_price_inr']
    .agg(['count', 'mean', 'median'])
    .sort_values('mean', ascending=False)
)

display(neighbourhood_analysis)
```

```python
size_analysis = (
    df.groupby('size_group', observed=True)['listing_price_inr']
    .agg(['count', 'mean', 'median'])
)

display(size_analysis)
```

```python
age_analysis = (
    df.groupby('age_group', observed=True)['listing_price_inr']
    .agg(['count', 'mean', 'median'])
)

display(age_analysis)
```

```python
ppsf_analysis = (
    df.groupby('neighbourhood')['price_per_sqft']
    .agg(['count', 'mean', 'median'])
    .sort_values('mean', ascending=False)
)

display(ppsf_analysis)
```

```python
print("Mean Price: ₹{:,.2f}".format(
    df['listing_price_inr'].mean()
))

print("Median Price: ₹{:,.2f}".format(
    df['listing_price_inr'].median()
))

print("Mean Property Area: {:.2f} sq.ft".format(
    df['property_area_sqft'].mean()
))

print("Mean Property Age: {:.2f} years".format(
    df['property_age_years'].mean()
))

print("Mean Price per Sq.Ft: ₹{:,.2f}".format(
    df['price_per_sqft'].mean()
))
```

```python
correlation = df[
    [
        'property_area_sqft',
        'property_age_years',
        'listing_price_inr',
        'price_per_sqft'
    ]
].corr()

print("Correlation Matrix:")
display(correlation)
```

```python
highest_neighbourhood = (
    neighbourhood_analysis['mean'].idxmax()
)

highest_price = (
    neighbourhood_analysis.loc[
        highest_neighbourhood, 'mean'
    ]
)

print("Highest Average Price Neighbourhood:",
      highest_neighbourhood)

print("Average Price: ₹{:,.2f}".format(
    highest_price
))
```

```python
highest_ppsf_neighbourhood = (
    ppsf_analysis['mean'].idxmax()
)

highest_ppsf = (
    ppsf_analysis.loc[
        highest_ppsf_neighbourhood, 'mean'
    ]
)

print("Highest Price/Sq.Ft Neighbourhood:",
      highest_ppsf_neighbourhood)

print("Average Price/Sq.Ft: ₹{:,.2f}".format(
    highest_ppsf
))
```

```python
output_file = '/content/cleaned_property_listings.csv'

df.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print(output_file)
```
